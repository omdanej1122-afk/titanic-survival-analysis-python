```python
## Outliers = Values are falls above maximum and below minimum (out of the range)
## Data more than 3 deviation and less than 3 deviation
```

```python
# Mean = Average of all values divided by total number of values
# Mode = frequently occuring value
# Median = Center of dataset
```

```python
df['Age'].median()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df.isnull().sum()
```

Now we are done with Age

```python
df.drop('Cabin',axis = 1) ## axis = 0 (row) , axis = 1 (column)
```

```python
df.columns
```

```python
df.drop('Cabin',axis = 1,inplace = True) ## For actually drop column
```

```python
df.columns
```

```python
df.isnull().sum()
```

we are done with Cabin

```python
df['Embarked'].mode() ## peoples are on boared form s(south)
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0]) ## Index is 0
```

```python
df.isnull().sum()
```

Now we are done with Embarked data is cleaned

```python
# Lets drop PassangerId , Name, Ticket as their is no role in analysis
df.drop(labels = ['PassengerId','Name','Ticket'],axis = 1,inplace=True)
```
